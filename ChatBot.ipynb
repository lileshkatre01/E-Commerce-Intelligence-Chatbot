```python
url = "https://raw.githubusercontent.com/DataScience75/Top_mentor_projects_Datasets/refs/heads/main/customer_reviews.csv"
import pandas as pd
df = pd.read_csv(url)
df.head()
```

```python
!pip install faiss-cpu
```

```python
import faiss
import numpy as np
from google import genai
```

```python
print(df.columns)
```

```python
document = []
for index, row in df.iterrows():
  text = f"""
  Product: {row['product_id']}
  Rating : {row['rating']}
  Review : {row['review_text']}
  """
  document.append(text)
```

```python
print(document[10])
```

```python
print(len(document))
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2") ##Embedding model
print(model)
```

```python
embeddings = model.encode(document)
embeddings.shape
```

```python
##Add Index in vector DB
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))
```

```python
##Retrivel Part
def retrieve_context(query,k = 3):
  query_embedding = model.encode([query])
  distance, indices = index.search(np.array(query_embedding),k = k)

  retrieve_docs = []
  for idx in indices[0]:
    retrieve_docs.append(document[idx])

  return "\n\n".join(retrieve_docs)
```

```python
res = retrieve_context("which product is worst ? ")
print(res)
```

```python
chat_history = []
def ask_question(question,top_k):

 global chat_history
 context =  retrieve_context(question,top_k)
 histroy = "\n".join(chat_history)

 prompt = f"""
 You are a Customer Review Analysis Assistant

 Conversation histroy :
 {histroy}
 Context :
 {context}
  Question :
  {question}

  Instruction;
  -Answer from the reviews only.
  -Summarize customer opinions.
  -Keep answer conversational.
  -Mention common issues if any
  """
 response = clt.models.generate_content(
      model = "gemini-3.6-flash",
      contents = prompt
  )
 answer = response.text
 chat_history.append(f'Question :{question}')
 chat_history.append(f'Answer :{answer}')
 return answer
```

```python
response = ask_question("Which product is best?",5)
print(response)
```
